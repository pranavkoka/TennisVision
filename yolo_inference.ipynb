```python
from ultralytics import YOLO
from pathlib import Path
import cv2
import time

input_image = Path(r'input\cv_image.png')
output_dir = Path("diff_models_output")
models = ["yolov8n.pt", "yolov8s.pt", "yolov8m.pt", "yolov8l.pt", "yolov8x.pt", "yolo26n.pt", "yolo26s.pt", "yolo26m.pt", "yolo26l.pt", "yolo26x.pt"]
output_dir.mkdir(parents=True, exist_ok=True)
```

```python
for model_name in models:

    print(f"\nRunning {model_name}...")
    model = YOLO(model_name)
    results = model.predict(
        source=str(input_image),
        verbose=False #doesn't give all the detailed terminal output
        )

    annotated_image = results[0].plot()
    model_short_name = Path(model_name).stem #.stem removes the .pt
    output_filename = f"{model_short_name}_{input_image.stem}.jpg"
    output_path = output_dir / output_filename
    cv2.imwrite(str(output_path), annotated_image)
    print(f"Saved: {output_path}")

print("\nDone!")
```

```python
WARMUP_RUNS = 5
BENCHMARK_RUNS = 30

for model_name in models:

    print(f"\nRunning {model_name}...")

    model = YOLO(model_name)

    # -------------------------
    # Warm-up
    # -------------------------
    for _ in range(WARMUP_RUNS):
        model.predict(
            source=str(input_image),
            verbose=False
        )

    # -------------------------
    # Benchmark
    # -------------------------
    times = []

    for _ in range(BENCHMARK_RUNS):

        start = time.perf_counter()

        results = model.predict(
            source=str(input_image),
            verbose=False
        )

        elapsed = time.perf_counter() - start
        times.append(elapsed)

    # -------------------------
    # Calculate average
    # -------------------------
    average_time = sum(times) / len(times)

    print(
        f"{model_name}: "
        f"{average_time:.4f} seconds average"
    )

print("\nDone once")

for model_name in models:

    print(f"\nRunning {model_name}...")

    model = YOLO(model_name)

    # -------------------------
    # Warm-up
    # -------------------------
    for _ in range(WARMUP_RUNS):
        model.predict(
            source=str(input_image),
            verbose=False
        )

    # -------------------------
    # Benchmark
    # -------------------------
    times = []

    for _ in range(BENCHMARK_RUNS):

        start = time.perf_counter()

        results = model.predict(
            source=str(input_image),
            verbose=False
        )

        elapsed = time.perf_counter() - start
        times.append(elapsed)

    # -------------------------
    # Calculate average
    # -------------------------
    average_time = sum(times) / len(times)

    print(
        f"{model_name}: "
        f"{average_time:.4f} seconds average"
    )

print("\nDone twice")
```
